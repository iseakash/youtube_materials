#### Strict Legal Rules (Must Follow)
- **Start with a letter or underscore**: a-z, A-Z, or an underscore, No numbers (e.g., 1variable is illegal).
- **Alphanumeric and underscores only**: Names can only contain letters, numbers, and underscores (a-z, 0-9, _).
- **No special characters**: You cannot use spaces, hyphens, or symbols like @, $, %, or !.
- **Case-sensitive**: Python treats capitalization differently. Age, age, and AGE are three completely different variables.
- **No Reserved Keywords**: You cannot use Python's built-in keywords (like if, else, while, class, def, return) as variable names.

    - **Reserved Keywords** in Python are hardcoded into Python's grammar. You are completely banned from reassigning them.
    - Reserved keywords are special words that Python has set aside for its own **internal syntax, logic, and structure**.
    - Because Python uses them to understand your code, you cannot use these words as **names for variables, functions, or classes**.

#### Assignment Operation on variables
- In Python, variable assignment does not copy data. Instead, the assignment operator (=) always copies memory references (pointers).
- When you write **variable = value**, Python creates the data object on the right side first, and then binds the variable name on the left side to that object's memory location.

#### Dynamic Typing in Python
- In statically-typed languages (like Java, C++, or Swift), types belong to the variable name.
- In Python, types belong to the data object itself.
- **Variables are Pointers, Objects have Types.** (Like a string attached to a balloon)

```python
# Create a pointer named 'box' to an integer object
box = 42
print(f"1. Pointer points to: {box} | Object Type: {type(box)} | Memory ID: {id(box)}")

# Retie the 'box' pointer to a string object
box = "Hello World"
print(f"2. Pointer points to: {box} | Object Type: {type(box)} | Memory ID: {id(box)}")

# Retie the 'box' pointer to a list object
box = [1, 2, 3]
print(f"3. Pointer points to: {box} | Object Type: {type(box)} | Memory ID: {id(box)}")
```

#### Four essential types of variable assignment behaviors
- **Multiple Assignment (Chained Assignment)**
    - You can assign a single value to multiple variables simultaneously in one line.
    - Because Python assigns by reference, all of these variables point to the exact same object in memory.
- **Simultaneous Assignment (Tuple Unpacking)**
    - Python allows you to assign multiple distinct values to multiple variables at the same time by separating them with commas.
- **The Classic Python Trick: Variable Swapping**
    - In traditional languages (like Java or C++), swapping the values of two variables requires creating a temporary third variable.
    - In Python, simultaneous assignment allows you to swap variables cleanly in a single line without using temporary storage variables.

```python
# Chained assignment
x = y = z = "Couch"

print(f"x id: {id(x)}")
print(f"y id: {id(y)}")
print(f"z id: {id(z)}")
print(f"Are they sharing memory? {x is y is z}")  # True

# If I change the value of any one variable, it will impact the id
x = "R"
print(f"x id: {id(x)}")
print(f"y id: {id(y)}")
print(f"z id: {id(z)}")
```

```python
# Assigning unique values to unique variables
item, weight_gm, is_required = "Paneer", 500, True

print(f"Item: {item}, Weight gm: {weight_gm}, Required: {is_required}")
```

```python
a = 5
b = 10
print(f"Before swap -> a: {a}, b: {b}")
print(f"Before swap -> id of a: {id(a)}, id of b: {id(b)}")

# The Python Swap Trick
a, b = b, a

print(f"After swap  -> a: {a}, b: {b}")
print(f"Before swap -> id of a: {id(a)}, id of b: {id(b)}")
```

## Some Advance Concepts:
- **Mutable vs. Immutable Objects**
    - Immutable Types: **int, float, str, tuple, bool**.
        - To alter, Python does not change the original data but points to a completely new object at a new memory address.
    - Mutable Types: **list, dict, set**.
        - You can add, remove, or modify elements within these objects without changing their address in memory.

- **Augmented Assignment Operators**
    - It combines a mathematical or logical operation with an assignment (e.g., +=, -=, *=, /=).
    - It modifies variables efficiently. However, memory behaves differently here depending on whether the object is mutable or immutable.
    
-  **Integer Caching (The Shared Memory Optimization)**
    - To save memory and speed up execution, Python uses a hidden optimization trick behind the scenes called **Integer Caching** (or integer reuse).
    - When Python starts up, it pre-loads an array of integers from -5 to 256 into memory. If you create any variables within this specific numerical range, Python will point them to the exact same pre-allocated memory addresses, instead of wasting RAM creating duplicate objects.

- **Automatic Garbage Collection**
    - In older languages like C, if you assign a variable to a piece of data and then reassign it, the old data sits around clogging up your computer's RAM until you manually delete it.
    - Python uses a system called **Reference Counting**. The moment a data object has zero variables pointing to it, Python automatically deletes it from your computer's memory.

```python
### Mutable vs. Immutable Objects

# --- IMMUTABLE EXAMPLE (Strings) ---
message = "Hello"
print(f"Original Address: {id(message)}")

# Attempting to modify the string
message = message + " World"
print(f"New Address:      {id(message)}") 
# Notice the IDs are different! Python created a whole new string object.

print("-" * 40)

# --- MUTABLE EXAMPLE (Lists) ---
numbers = [1, 2, 3]
print(f"Original Address: {id(numbers)}")

# Modifying the list in place
numbers.append(4)
print(f"New Address:      {id(numbers)}") 
# Notice the IDs are identical! The original object was modified directly.
```

```python
### Augmented Assignment Operators

# --- Immutable Augmented Assignment ---
num = 10
print(f"Initial int memory ID: {id(num)}")
num += 5  # Creates a brand new integer (15) and changes the nametag pointer
print(f"Updated int memory ID: {id(num)}")  # ID changes

print("-" * 40)

# --- Mutable Augmented Assignment ---
my_list = [1, 2]
print(f"Initial list memory ID: {id(my_list)}")
my_list += [3, 4]  # Modifies the original list directly in place (like .extend())
print(f"Updated list memory ID: {id(my_list)}")  # ID stays identical
```

```python
result = 1
for i in range(1,4):
    result += 1
    print("result:", result, "memory ID:", id(result))

result = []
for i in range(1,4):
    result.append(i)
    print("result:", result, "memory ID:", id(result))
```

```python
### Integer Caching (The Shared Memory Optimization)

# --- WITHIN THE CACHE RANGE (-5 to 256) ---
x = 100
y = 100
print("Inside Cache Range (100):")
print(f"x memory ID: {id(x)}")
print(f"y memory ID: {id(y)}")
print(f"Do x and y share the exact same memory? {x is y}") # True

print("-" * 40)

# --- OUTSIDE THE CACHE RANGE ---
a = 300
b = 300
print("Outside Cache Range (300):")
print(f"a memory ID: {id(a)}")
print(f"b memory ID: {id(b)}")
print(f"Do a and b share the exact same memory? {a is b}") # False
```

```python
### Garbage Collection Visualization
data = "Important Data"  # Reference count = 1

data = "New Data"  
# The string "Important Data" now has a reference count of 0.
# Python automatically deletes "Important Data" from your RAM behind the scenes!
print(data)
```

```python
import sys

# Create a list (1 reference created)
my_plate = ["Gulab Jamun", "Samosa"]

# sys.getrefcount() temporarily points to the object to count it, 
# so the output will always be (Actual Count + 1).
print("Initial reference count:", sys.getrefcount(my_plate) - 1) # Expected: 1

# Add a second nametag pointing to the same list
friend_plate = my_plate
print("Count after sharing:", sys.getrefcount(my_plate) - 1)    # Expected: 2

# Peel off one nametag
friend_plate = None
print("Count after friend leaves:", sys.getrefcount(my_plate) - 1) # Expected: 1
```
