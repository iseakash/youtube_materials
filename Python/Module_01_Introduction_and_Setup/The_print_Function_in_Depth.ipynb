#### 1. Positional Arguments: *objects
- The asterisk (*) indicates that print() can accept any number of positional arguments (zero or more).

```python
print("Hello", 42, [1, 2, 3], True, 400, 500, "string", "...", 5.876906)
```

```python
# Code Cell: Unpacking Collections
matrix_row = ["Feature_1", "Feature_2", "Label"]

# Standard print shows the collection bracket wrappers
print(matrix_row) 

# Unpacked print feeds elements as individual positional arguments
print(*matrix_row)

# Clean custom formatting combined with separation parameter
print(*matrix_row, sep=" | ")
```

```python
# Code Cell: Efficient Debug Logging
epoch = 14
batch_idx = 400
loss_val = 0.02315

# Highly readable, memory efficient, no string addition required
print("➔ [TRAIN]", "Epoch:", epoch, "Batch:", batch_idx, "Loss:", loss_val)

# The Bad Way: Forcing manual type conversions and string additions (print("Loss: " + str(loss) + " at step " + str(step)))
# This creates temporary string objects in your computer's memory, slowing down heavy data loops.
```

```python
# Visualizing a 3x3 Confusion Matrix or Grid
grid = [[1, 30, 96], [10, 3, 96], [1, 3, 96]]

print("--- Confusion Matrix Layout ---")
for row in grid:
    # print(row)
    # Unpacks the row array into 3 separate arguments, formatted with tabs
    print(*row, sep="\t")
```

```python
# Functional Pipeline Reporting
def compute_summary_statistics():
    # Imagine this pulls varying metrics based on dataset size
    return "Mean: 4.5", "StdDev: 1.2", "Variance: 1.44"

print(compute_summary_statistics())
# Pipes the returned tuple items directly into individual arguments
print(*compute_summary_statistics(), sep=", ")
```

### 2. Keyword Arguments (The Hidden Power)
#### **sep (Separator):**
- By default, Python places a single space between multiple objects. You can change this behavior using the sep argument.

```python
# Default behavior (space)
print("Python", "is", "fun")
```

```python
# Custom separator: Hyphen
print("01", "01", "2026", sep="-")
```

```python
# Custom separator: Newline character
print("Apple", "Banana", "Cherry", sep="\n")
```

#### **end (End of Line):**
- print() appends a character to the end of the text (By default: newline i.e., \n)

```python
# Default behavior (moving to next line)
print("Hello")
print("World")
```

```python
# Custom end: Space instead of newline
print("Hello", end=" ")
print("World")
```

```python
# Custom end: Creating a loading bar effect
print("Loading", end="...")
print(" Done!")
```

##### Inline Rewriting (end="\r")

How It Works Under the Hood?

* \r (Carriage Return): \r moves the blinking text cursor back to column 0 of the current line.
* The Overwriting Effect: When you print a new string after \r, Python does not delete the old text. It simply types over it, character by character.
* The Ghost Text Trap (Crucial Developer Tip): If your new string is shorter than your old string, the leftover characters from the old string will still be visible!

Real-World Uses
- Monitoring Long-Running Machine Learning Epochs
- Tracking Large ETL Data Pipelines

```python
# Clear and Rewrite the Same Line
import time

for percent in range(0, 101, 20):
    # \r moves the cursor back to the start; flush forces the screen to update
    print(f"Downloading dataset... {percent}%", end="\r", flush=True)
    time.sleep(0.5)
print("\nDownload Complete!") # Newline at the end so the next print doesn't overwrite it
```

```python
import time

for percent in range(9, 0, -1):
    # \r moves the cursor back to the start; flush forces the screen to update
    print(f"Downloading dataset... {[percent]*percent}%", end="\r", flush=True)

    # Code Cell: The Professional Fix
    # Adding spaces to clear out any old text remnants
    print(f"Downloading dataset... {[percent]*percent}%                 ", end="\r", flush=True) 
    time.sleep(0.5)
print("\nDownload Complete!") # Newline at the end so the next print doesn't overwrite it
```

#### **file - sys.stdout** (Redirecting Output)
- By default, print() sends text to sys.stdout (your screen/notebook screen).
- You can pass a file object to this argument to write data directly into a file instead.

```python
print("This text is saved to a file, not the screen.")
```

```python
with open("output1.txt", "w") as f:
    print("This text is saved to a file, not the screen.", file=f)
```

#### **flush (Clearing the Buffer)**
- Python's output streams are usually buffered, meaning they wait until they collect enough data before writing to the screen.
- Setting flush=True forces Python to clear the buffer and display the text instantly.

```python
import time
import sys

print("--- TESTING WITHOUT FLUSH ---")
for i in range(10):
    print(".", end="")  # No newline, data sits in the memory buffer
    time.sleep(0.5)     # You will see NOTHING for 2.5 seconds, then all dots appear at once
print("\nDone")

print("--- TESTING WITH FLUSH ---")
for i in range(10):
    print(".", end="", flush=True)  # Forces data out of memory instantly
    time.sleep(0.5)                 # You will see one dot appear every 0.5 seconds
print("\nDone")
```

### Bonus Tips: Printing with Color: ANSI Escape Codes
- ANSI escape codes are invisible instructions telling terminal emulator to change its physical display settings before rendering the text.
#### How It Works Under the Hood?
- **\033 (The Escape Character):** This initiates the sequence. It alerts the system that what follows is an internal formatting command, not literal text to print.
- **[ (Control Sequence Introducer):** Signals the start of the configuration parameters.
- **The Numbers (The Parameters):** These are separated by semicolons and dictate the style:
    * **Style** (1): Bold or high intensity (0 is normal, 4 is underlined).
    * **Text Color** (31): Red (32 is green, 34 is blue, 37 is white).
    * **Background Color** (40): Black (41 is red, 42 is green, 46 is cyan).
- **m (The Terminator):** Tells the terminal that the configuration setup is finished and the actual text to be printed is coming next.
- **\033[0m (The Reset Switch):** Crucial! This resets the terminal back to its default system styling.
    * If you forget this, every single print statement after this line in your notebook will stay colored forever!

Note: ANSI codes do not require all three parameters.

#### Real-World Uses for ANSI Color Codes
- Instant Visual Triage in CLI Tools & Logs (Green: Success, Yellow: Warnings, Red: Error)
- Highlighting Discrepancies during Code Debugging

```python
# Code Cell# Code Cell
# Formula: \033[STYLE;COLOR;BACKGROUNDm YOUR TEXT \033[0m
print("\033[1;31;46m Critical Error1!")
print("\033[1;31;40m [CRITICAL ERROR] \033[0m Connection to database lost.")
print("\033[2;32;40m [SUCCESS] \033[0m Model trained in 4.2 seconds.")
print("\033[4;41m Informational Message \033[0m: Learning print in depth")
```

```python
# Code Cell: Proving Parameter Independence

# 1. No Style parameter: Default text thickness, Red Background
print("\033[41m Background Only \033[0m")

# 2. No Background parameter: Underlined text, Red Text Color
print("\033[4;31m Underline & Red Text \033[0m")

# 3. Order doesn't matter: Terminal parses 41 as background and 1 as style regardless of sequence
print("\033[41;1m Bold text on Red Background \033[0m")
```

```python
# Example 01: Showing the disaster of forgetting the reset switch
print("\033[1;31mThis is red text...")
print("Oops! I forgot the reset switch, so this text is accidentally red too!")
print("\033[0mAh, fixed it. This is back to standard terminal color.")

# Example 02: Showing the disaster of forgetting the reset switch
print("\033[1;31;46m Critical Error1!")
print("Critical Error2!")
print("Critical Error3! \033[0m")
print("Critical Error4!")
```
