### How input() works?
1. **The Pause**: Python sends a command to the operating system to temporarily freeze your program's execution. It tells the CPU to stop processing your code and wait.

2. **The Buffer**: As you type characters on your keyboard, the operating system catches them and stores them in a temporary storage area called a text buffer.

3. **The Trigger**: The operating system does not give anything back to Python until you press the Enter key. Pressing Enter sends a special line-break character (\n) to the buffer, which signals that you are done typing.

4. **The Cleanup**: Python wakes up, grabs the text from the buffer, automatically strips away the hidden \n character from the end, and packages the remaining text into a brand-new string object.

### two or more input()

```python
print("Type 'exit' or 'quit' to end the chat.")

while True:
    # 1. Ask for input every turn
    user_message = input("You: ")
    
    # 2. Check for the exit condition (converted to lowercase for safety)
    if user_message.lower() == "exit" or user_message.lower() == "quit":
        print("Goodbye!")
        break # Exits the loop immediately
        
    # 3. Otherwise, respond to the input
    print(f"Bot: You just said '{user_message}'")
```

### Combining Multiple Inputs into One Sentence
