## What are Booleans?
- Computers operate on binary (1s and 0s).
- Python implements True and False as human-readable labels directly wired to these binary states.
- Key Characteristics:
    - Strict Capitalization: Always capitalize the first letter (True and False).
    - Integer Subclass: True acts like the number 1, and False acts like 0.

```python
is_sunny = True
is_raining = False
print(type(is_raining))
```

```python
x = True
t5 = True
print(id(x))
print(id(t5))
```

```python
print(isinstance(True, int))  # True
print(True + True)            # Returns: 2
print(False * 50)             # Returns: 0

# Python literally maps them to 1 and 0 in memory tables
print(True == 1)              # True
```

## How to Use - Comparison Expressions & Logical Operators
- Comparison Expressions:
    - When Python encounters operators like >, <, or ==, it evaluates the expressions on both sides and returns a single Boolean object directly from memory.
    - Example: Security guard checking IDs at a club gate
- Logical Operators: 
    - Can combine multiple Boolean conditions using logical operations
    - and: Yields True only if every single statement is true
    - or: Yields True if at least one statement is true
    - not: Reverses the value (not True becomes False)

```python
age = 25
can_enter = age >= 21
print(can_enter)
```

```python
print(10 == 9)
```

```python
# Both expressions return True, but they point to the exact same memory address
print(id(5 > 3)) 
print(id(10 == 10)) 

# This will output the exact same number for both
```

```python
# Logical Operators: can combine multiple Boolean conditions using logical operations
is_adult = False
has_ticket = True

print(is_adult and has_ticket)
print(is_adult or has_ticket)
```

```python
not False
```

## The bool() Function
- Python does not require variables to be strict Booleans inside conditional statements.
- Every built-in object inherently maps to a True or False state based on its "substance" or magnitude.

```python
bool("")
```

## Basic Truthy/Falsy
- When Python evaluates non-Boolean objects (like numbers or lists) in logical conditions, it uses a concept called **Truthiness**.
- To understand why Python does this, think of data as physical matter:
    - **Falsy or Falsiness** values represent **Nothingness**, Emptiness, or Dead States.
        - 0 (Zero): The absence of quantity.
        - "" (Empty String): A text box with nothing inside.
        - None : Literally "nothing here" or a blank space.
        - [] or {}: The container exists, even if it is currently empty.
        - False: The ultimate definition of non-occurrence.
    - **Truthy or Truthiness** values represent **Existence**, Substance, or Active States.
        - 1 or -5: A quantity exists.
        - "false" (String): The letters exist, even though the word spells false.
        - [False] or {None}: The container exists, even if it is currently empty.
        - True: The ultimate definition of non-occurrence.

```python
print(bool(0))     # No magnitude
print(bool(42))    # Has magnitude

print(bool(""))    # Empty string
print(bool("Hi"))  # Contains text

print(bool(None))  # Represents nothingness
```

```python
# The Rule of or: It looks for the first Truthy value it can find.
# The moment it finds one, it stops evaluating (short-circuits) and hands you that exact object.

user_input = ""
default_name = "Guest1"
active_name = user_input or default_name or "Akash"
print(active_name)
```

```python
# The Rule of and: It requires everything to be Truthy.
# It evaluates from left to right. If it hits a Falsy value, it stops immediately and returns it.
# If it makes it all the way to the end without failing, it returns the very last object that confirmed the success.

print([1, 2] and "Hello" and "Akash")
```
