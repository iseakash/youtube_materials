### Overview of built-in Numeric data types

Python supports three built-in numeric data types: int, float, and complex.


| Numeric Type | Description | Example Syntax | Common Use Case |
| :--- | :--- | :--- | :--- |
| **int** | Whole numbers of unlimited length. | `x = 42` | Counting, indexing, loops. |
| **float** | Numbers containing fractions/decimals. | `y = 3.14` | Measurements, financial prices. |
| **complex** | Numbers with real and imaginary parts. | `z = 2 + 3j` | Physics, signal processing. |


### Deep dive into each Numeric data types

1. **Integers (int)**

- Basic Concept:
    - Represent positive or negative whole numbers without any decimal points.
    - **Readability**: You can use underscores as visual digit separators (e.g., 1_000_000) to make large numbers easier to read.

- Advance Concept:
    - **Unlimited Precision**: In Python, integers can grow as large as your system memory allows without overflowing.
    - **Alternative Bases**: You can define integers in binary (0b), octal (0o), or hexadecimal (0x) formats.

```python
count = 105
large_number = 999_999_999_999_999  # Allowed without overflow
binary_val = 0b1010                  # Evaluates to 10

# Note: '_' in 10_000 is a visual separator introduced in Python 3.6. It functions exactly like a comma in written numbers (10,000).
```

```python
# Convert the base-10 number 42 into base-2 binary
print(bin(42))

# Output: '0b101010' 
# (The '0b' means "this is a binary number")
# Computers do not use base-10 math (0-9) like humans; they use base-2 binary (0 and 1) powered by tiny on/off electronic switches.
```

```python
import sys

small_num = 123
large_num = 999_999_999_999_999
huge_num  = 10**100  # A googol (1 followed by 100 zeros)

print(f"Small Num Memory: {sys.getsizeof(small_num)} bytes")  # Output: ~28 bytes
print(f"Large Num Memory: {sys.getsizeof(large_num)} bytes")  # Output: ~32 bytes
print(f"Huge Num Memory:  {sys.getsizeof(huge_num)} bytes")   # Output: ~72 bytes
```

```python
# New 2026/Modern Python Catch: String Conversion Limits
# Python will block you if the number has more than 4,300 digits - The Security Threat: Denial of Service (DoS)

# This math runs perfectly in milliseconds
massive_calc = 10**4300
print(f"Massive Num Memory:  {sys.getsizeof(massive_calc)} bytes")

# This line will crash with a ValueError to protect your CPU
print(massive_calc) 
# Error: ValueError: Exceeds the limit (4300) for integer string conversion
```

```python
# Option A: Increase the limit to 10,000 digits
sys.set_int_max_str_digits(10000)

# Option B: Remove the limit entirely (set to 0)
sys.set_int_max_str_digits(0)

# Now this 5,000-digit conversion will print perfectly
massive_calc = 10**5000
print(massive_calc)
```

```python
### Integer unexpected behaviors 
# Memory Consumption: Python integers scale dynamically, but very large numbers consume massive amounts of system memory and slow down execution.
# This runs perfectly fine but uses extra memory overhead
huge_int = 10 ** 10_000_000
```

2. **Floating-Point Numbers (float)**

- Basic Concept:
    - **Decimal Points**: Any number written with a decimal point automatically becomes a float.
    - **Scientific Notation**: You can append an e or E to represent powers of 10.

- Advance Concept:
    - **Precision Limit**: Unlike integers, floats are constrained by standard 64-bit hardware limitations, meaning minor rounding limits apply to extremely precise fractions.

```python
price = 19.99
distance = -0.5
scientific = 35e3  # Evaluates to 35000.0
```

```python
# Computers represent base-10 decimals in base-2 binary, certain fractions cannot be stored exactly.

print(0.1 + 0.2 == 0.3)
print(0.1 + 0.2)
```

```python
# Check how the computer actually stores 0.1
print((0.15).as_integer_ratio())
```

```python
### Floating-Point Overflow
# While integers have infinite precision, floats have a strict maximum size boundary.
# Note: Computer's operating system and processor have hardware-defined limits for how many base-10 digits they can reliably track before turning them into base-2 approximations

import sys

# Print system rules for floating-point math
print(sys.float_info)
print("float_info.max:", sys.float_info.max)

# Going beyond this threshold triggers an overflow
huge_float = 2.0 ** 1024   # Raises: OverflowError: 'long int too large to convert to float'
```

```python
### Special Floating-Point Constants
# Floats include three special values (inf, -inf, and nan) that alter standard math logic. They are case-insensitive when parsed from strings.
infinity = float('inf')
print(infinity > 10**100)  # Returns True
print(infinity + 5)        # Returns inf
print(infinity - infinity) # Returns nan (Not a Number)
```

```python
# Not a Number (nan)
not_a_num = float('nan')
print(not_a_num == not_a_num)  # Returns False!

# The Fix: Use math.isnan() to verify its identity safely
import math
print(math.isnan(not_a_num))   # Returns True
```

```python
# Contradicting Infinity
print(float('inf') - float('inf'))  # Outputs: nan
print(float('inf') / float('inf'))  # Outputs: nan
```

3. **Complex Numbers (complex)**
- Basic Concept:
    - Complex numbers consist of a real part and an imaginary part.
    - **The j Suffix**: Python uses the literal j (or J) to denote the imaginary unit (sqrt(-1)) instead of the mathematical i.
    - **Component Access**: You can isolate the distinct parts of a complex number using .real and .imag attributes.
    - **Use Cases**: These are highly specialized for scientific applications like electrical engineering, geometry, and fluid dynamics.

```python
num = 3 + 4j
print(num.real)  # Outputs 3.0
print(num.imag)  # Outputs 4.0
```

```python
num = -0.0-0j
print(num.real)
print(num.imag)
```

```python
# The Invisible 0.0 (A Trapping Bug) As 0.0 and -0.0 are two completely different binary strings
z = 3j
print(z.real)  # Output: 0.0 (It actively occupies memory)
```

```python
# Complex Numbers Cannot Be Compared

# Python allows you to check for exact structural equality:
print((2 + 3j) == (2 + 3j))  # Outputs: True
print((3 + 2j) == (2 + 3j))  # Outputs:
```

```python
z1 = 2 + 3j
z2 = 1 + 5j

# This will crash the notebook
print(z1 > z2)  
# Output: TypeError: '>' not supported between instances of 'complex' and 'complex'
```

```python
print(abs(z1), abs(z2))
```

```python
# Comparing by Magnitude (Distance from Zero)

# Use the built-in abs() function to calculate the scalar magnitude
print(abs(z1) > abs(z2))  # Outputs: False (Checks which vector is physically longer)
```

```python
# Comparing Component-by-Component

# Explicitly isolate the components
if z1.real == z2.real:
    print(z1.imag > z2.imag)
else:
    print(z1.real > z2.real)  # Outputs: True (2 is greater than 1)
```

```python

# The Int Box (int)

# [--- Reference Count ---] -> 8 bytes
# [---- Type Pointer -----] -> 8 bytes
# [----- Object Size -----] -> 8 bytes
# [--- Actual Data Slot --] -> 4 bytes
# -------------------------------------
# TOTAL                     = 28 bytes
```

```python

# The Float Box (float)

# [--- Reference Count ----] -> 8 bytes
# [---- Type Pointer ------] -> 8 bytes
# [--- 64-Bit Float Value -] -> 8 bytes
# -------------------------------------
# TOTAL                     = 24 bytes
```

```python

# The Complex Box (complex)

# [--- Reference Count ---] -> 8 bytes
# [---- Type Pointer -----] -> 8 bytes
# [--- 64-Bit Real Float --] -> 8 bytes
# [--- 64-Bit Imag Float -] -> 8 bytes
# -------------------------------------
# TOTAL                     = 32 bytes
```

```python

# An Integer is like a train while A Float or A Complex is like a standard shipping container.
# Both float and complex use different internal boxes because they do not change size dynamically like integers do.
# Integers are flexible and infinite, while floats and complex numbers are rigid and fixed.
```


## Type Conversion (Casting) & Mixed Arithmetic
Python allows you to change a number's data type either automatically or deliberately.
- **Implicit Widening**: When mixing types in mathematical operations, Python automatically converts the narrower type to the broader type *(int → float → complex)* to preserve precision.
- **Explicit Casting**: Use constructors to force a specific type. Note that casting a float to an int will truncate *(chop off)* the decimal digits entirely.

```python

# Mixed Arithmetic
result = 5 + 2.0 + (1+3j)  # Returns 7.0 (float)
result
```

```python

# Mixed Arithmetic
result = (1+3j) - (3j) # Returns (1+0j) (complex)
result
```

```python

# Explicit Constructors
to_int = int(5.9)       # Returns 5 (drops decimals)
to_float = float(10)     # Returns 10.0
to_complex = complex(4)  # Returns (4+0j)
```

```python

# Truncation vs. Rounding
print(int(5.99))   # Returns 5
print(int(-5.99))  # Returns -5
# int() (Truncation): Always rounds towards zero

print(round(5.99))   # Returns 6
print(round(-5.99))  # Returns -6
# round() rounds: Moves to the nearest whole number or nearest even number
```

```python

z = 3 + 4j
float(z)  # Raises: TypeError: can't convert complex to float
```

```python

print(round(2.5))  # Returns 2
print(round(3.5))  # Returns 4
```
